```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from quantificationlib.baselines.cc import CC, PCC
from quantificationlib.baselines.ac import AC, PAC
from quantificationlib.multiclass.df import HDy
from quantificationlib.multiclass.em import EM

import matplotlib.pyplot as plt

import sys, os
sys.path.append(os.path.abspath(".."))

from src.data.bags import generate_bags

from aeon.classification.deep_learning import FCNClassifier, LITETimeClassifier
```

```python
data = np.load("./data/processed/mitdb_preprocessed.npz")

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]
rec_train = data["rec_train"]
rec_test = data["rec_test"]

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("Train positives:", np.sum(y_train==1))
print("Test positives:", np.sum(y_test==1))
```

```python
bags_X_test_100, bags_prev_test_100 = generate_bags(
    X_test, y_test,
    bag_size=100,
    n_bags=2000,
    random_state=42
)

bags_X_test_200, bags_prev_test_200 = generate_bags(
    X_test, y_test,
    bag_size=200,
    n_bags=2000,
    random_state=42
)

bags_X_test_500, bags_prev_test_500 = generate_bags(
    X_test, y_test,
    bag_size=500,
    n_bags=2000,
    random_state=42
)
```

```python
results = {
    'bag_id': [],
    'bag_size': [],
    'method': [],
    'classifier': [],
    'mae': [],
    'pred_pos': [],
    'true_pos': [],
}
```

## CC LITETime Predictions

```python
estimator = LITETimeClassifier(random_state=42, save_best_model=True)
# estimator.fit(X_train, y_train)

cc = CC(estimator_test=estimator)
cc.fit(X_train, y_train)
```

```python
for idx, (Xb, prev_true) in enumerate(zip(bags_X_test_100, bags_prev_test_100)):
    Xb_f = Xb.reshape(len(Xb), -1)
    prev_pred = cc.predict(Xb_f)

    results['bag_id'].append(idx)
    results['bag_size'].append(100)
    results['method'].append('CC')
    results['classifier'].append('LITETime')
    results['mae'].append(abs(prev_pred[1] - prev_true))
    results['pred_pos'].append(prev_pred[1])
    results['true_pos'].append(prev_true)

for idx, (Xb, prev_true) in enumerate(zip(bags_X_test_200, bags_prev_test_200)):
    Xb_f = Xb.reshape(len(Xb), -1)
    prev_pred = cc.predict(Xb_f)

    results['bag_id'].append(idx)
    results['bag_size'].append(200)
    results['method'].append('CC')
    results['classifier'].append('LITETime')
    results['mae'].append(abs(prev_pred[1] - prev_true))
    results['pred_pos'].append(prev_pred[1])
    results['true_pos'].append(prev_true)

for idx, (Xb, prev_true) in enumerate(zip(bags_X_test_500, bags_prev_test_500)):
    Xb_f = Xb.reshape(len(Xb), -1)
    prev_pred = cc.predict(Xb_f)

    results['bag_id'].append(idx)
    results['bag_size'].append(500)
    results['method'].append('CC')
    results['classifier'].append('LITETime')
    results['mae'].append(abs(prev_pred[1] - prev_true))
    results['pred_pos'].append(prev_pred[1])
    results['true_pos'].append(prev_true)
```

```python
results_df = pd.DataFrame(results)
results_df.to_csv('../results/baselines_2.csv')
```

```python
results_df.query('method == "CC" and bag_size == 200').mae.mean(), results_df.query('method == "CC" and bag_size == 200').mae.std()
```
